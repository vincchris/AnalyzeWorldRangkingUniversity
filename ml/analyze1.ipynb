```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('../data/2026 QS World University Rankings.csv')
```

```python
df = df.dropna(subset=['Status'])
df.head()
```

```python
df.isnull().sum()
```

```python
df['Previous Rank'] = df['Previous Rank'].fillna(df['Previous Rank'].mean())
df['IFR SCORE'] = df['IFR SCORE'].fillna(df['IFR SCORE'].mean())
df['IFR RANK'] = df['IFR RANK'].fillna(df['IFR RANK'].mean())
df['ISR SCORE'] = df['ISR SCORE'].fillna(df['ISR SCORE'].mean())
df['ISR RANK'] = df['ISR Rank'].fillna(df['ISR Rank'].mean())
df['ISD SCORE'] = df['ISD SCORE'].fillna(df['ISD SCORE'].mean())
df['ISD RANK'] = df['ISD Rank'].fillna(df['ISD Rank'].mean())
df['IRN SCORE'] = df['Previous Rank'].fillna(df['ISR Rank'].mean())
df['IRN RANK'] = df['Previous Rank'].fillna(df['ISR Rank'].mean())
df['SUS SCORE'] = df['Previous Rank'].fillna(df['ISR Rank'].mean())
df['SUS RANK'] = df['Previous Rank'].fillna(df['ISR Rank'].mean())
df.isnull().sum()
```
